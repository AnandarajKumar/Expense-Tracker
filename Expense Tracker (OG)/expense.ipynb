```python
from faker import Faker
import random
```

```python
fake = Faker()
```

```python
categories = ["Food", "Transportation", "Bills", "Entertainment", "Groceries"]
payment_modes = ["Card","UPI","Cash"]
```

```python
expense = {
    "Date" : fake.date_this_month(),
    "Category" : random.choice(categories),
    "payment mode" : random.choice(payment_modes),
    "Description" : fake.sentence(),
    "Amount" : round(random.uniform(10,1000),2)
}
print(expense)
```

```python
import pandas as pd
```

```python
def gen_exp_data(num_entries = 100) :
    data = []
    for _ in range(num_entries) :
        expense = {
    "Date" : fake.date_this_month(),
    "Category" : random.choice(categories),
    "payment mode" : random.choice(payment_modes),
    "Description" : fake.sentence(),
    "Amount" : round(random.uniform(10,1000),2)
        }
        data.append(expense)
        return pd.DataFrame(data)
        
exp_data = gen_exp_data(100)
```
